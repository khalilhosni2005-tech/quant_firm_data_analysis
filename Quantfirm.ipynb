```python
import pandas as pd
import numpy as np
```

```python
data = {
    "Employee_ID": [
        101, 102, 103, 104, 105, 106,
        107, 108, 109, 110, 111, 112
    ],
    "Name": [
        "Alex", "Sarah", "Youssef", "Emma",
        "Daniel", "Lina", "Adam", "Sofia",
        "James", "Maya", "Omar", "Leo"
    ],
    "Company": [
        "Jane Street", "Jane Street", "HRT", "HRT",
        "Citadel", "Citadel", "Jane Street", "HRT",
        "Citadel", "Jane Street", "HRT", "Citadel"
    ],
    "Department": [
        "Trading", "Research", "Trading", "Engineering",
        "Research", "Trading", "Engineering", "Research",
        "Engineering", "Trading", "Research", "Trading"
    ],
    "Salary_USD": [
        150000, 125000, 180000, 110000,
        140000, 160000, 115000, 135000,
        120000, 145000, 130000, 155000
    ],
    "Age": [
        26, 24, 29, 27, 31, 28,
        25, 30, 26, 29, 27, 32
    ],
    "Years_Experience": [
        3, 2, 5, 4, 6, 4,
        2, 5, 3, 4, 3, 7
    ],
    "Remote": [
        True, True, False, True,
        False, True, True, False,
        False, True, False, True
    ]
}
```

```python
df = pd.DataFrame(data)
df
```

```python
#level 1

#first five rows of the dataset
print(df.head(5))
#last three rows of the dataset
print(df.tail(3))
#number of rows and columns
print(df.shape)
#every column type
print(df.dtypes)
#descriptif statistics for salary_usd
print(df['Salary_USD'].describe())
```

```python
#level 2

#employee of index 4
print(df.iloc[4])
#salary of the employee at index 4
print(df['Salary_USD'].loc[2])
#employees earning more than $140000
print(df[df['Salary_USD'] > 140000])
#employees earn more than $130000 and age under 30
print(df[(df['Salary_USD'] > 130000) & (df['Age'] < 30)])
#employees working in research or engineering
print(df[(df['Department'] == 'Research') | (df['Department'] == 'Engineering')])
#name and salaries of employeeswho works remotly and have atleast three years of experience
print(df[(df['Remote'] == True) & (df['Years_Experience'] >= 3)][['Name', 'Salary_USD']])

```

```python
#level 3

#Add Annual_Bonus column
df['Annual_Bonus'] = df['Salary_USD'] * 0.1
#add total_Compensation column
df['Total_Compensation'] = df['Salary_USD'] + df['Annual_Bonus']
#insert the seniority column
seniority_list= []
for i in df['Years_Experience']:
    if i >= 5:
        seniority_list.append('Senior')
    else:
        seniority_list.append('Junior')
df.insert(2, 'Seniority', seniority_list)
#add monthly salary using assign
new_df = df.assign(monthly_salary = df['Salary_USD']/12)
#copy of df without remote column
copy_df = df.drop('Remote', axis=1)
#remove employee with the index 5
copy_df.drop(5, inplace=True)



```

```python
#level 4

#number of employees in each company
count = df.groupby('Company')['Employee_ID'].count()
print(count)
#avg salary of each company
avg_salary = df.groupby('Company')['Salary_USD'].mean()
print(avg_salary)
#max salary of each departement
max_salary = df.groupby('Department')['Salary_USD'].max()
print(max_salary)
#how many employees in each depatment
employee_count = df.groupby('Department')['Employee_ID'].count()
print(employee_count)
#groupby company and department
employee_count = df.groupby(['Company', 'Department'])['Salary_USD'].mean()
print(employee_count)
```

```python
#level 5

#all unique company names
print(df['Company'].unique())
#compnies employees number using value_count()
print(df['Company'].value_counts())
#sort employes by salary from the highest to the lowest
df['Salary_USD'].sort_values(ascending=False)
#sort employes by there alphabeticly by name
df['Name'].sort_values()
#series containing only unique salaries
series_list = df['Salary_USD'].unique()
series = pd.Series(series_list)
print(series)
#last question of level 5
idk = pd.DataFrame({"1": 1, "2": np.nan, "3": 3}, index=[1])
idk.fillna(value=2, inplace=True)

```

```python
#level 6
#company with the most employees
idk = df.groupby('Company')['Employee_ID'].count()
print(idk[idk == idk.max()])
#department with the highest average salary
idk = df.groupby('Department')['Salary_USD'].mean()
print(idk[idk == idk.max()])
#three highest paid employees
print(df[['Name', 'Company', 'Salary_USD']].sort_values(by='Salary_USD', ascending=False).head(3))
#avg salary of remote workers + comparison
remote_avg = df.groupby('Remote')['Salary_USD'].mean()
print(f"the avg salary of remote workers is : {remote_avg[True]}")
if remote_avg[True] > remote_avg[False]:
    print("remote workers get paid more than non remote worker")
else:
    print("non remote workers get paid more than remote worker")
#new data frame + percentage salary column
ndf = df[df['Salary_USD'] >= 130000].copy()
avg = df['Salary_USD'].mean()
ndf['percentage'] = (ndf['Salary_USD'] * 100)/avg
#Sumary
print("""
Jane Street, HRT, and Citadel each employ four people.
Trading has the highest average salary at $158,000.
The highest-paid employee is Youssef, earning $180,000.
Remote employees average approximately $137,143, compared with $141,000 for non-remote""")

```
